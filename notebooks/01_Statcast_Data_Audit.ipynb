## 1. Data Collection

Statcast provides pitch-level information on MLB pitchers, including release characteristics, pitch movement, spin, velocity, and pitch outcomes.

We will use the `pybaseball` Python library to retrieve this information directly.

For our initial test, we will collect Sean Manaea's pitching data over a two-week period during the 2024 season.

This small sample will allow us to verify the data retrieval process and inspect the available variables before developing an automated collection pipeline.


```python

import pandas as pd
from pybaseball import statcast_pitcher

# Initial pitcher and date range
PITCHER_ID = 640455
START_DATE = "2024-07-01"
END_DATE = "2024-07-15"

# Retrieve pitch-level Statcast data
data = statcast_pitcher(
    START_DATE,
    END_DATE,
    player_id=PITCHER_ID
)

# Display basic dataset information
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])

display(data.head())
```
